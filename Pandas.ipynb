```python
import pandas as pd
import numpy as np
```

```python
energy = pd.read_excel('Energy Indicators.xls', skiprows=17, skipfooter=38)

energy = energy.drop(energy.columns[[0, 1]], axis=1)

energy.columns = ['Country', 'Energy Supply', 'Energy Supply per Capita', '% Renewable']
energy.replace('...', np.nan, inplace=True)

energy['Energy Supply'] = energy['Energy Supply'] * 1000000
energy['Country'] = energy['Country'].str.replace(r'\d+', '', regex=True)
energy['Country'] = energy['Country'].str.replace(r' \(.*\)', '', regex=True)
energy['Country'] = energy['Country'].str.strip()

dict_names = {
    "Republic of Korea": "South Korea",
    "United States of America": "United States",
    "United Kingdom of Great Britain and Northern Ireland": "United Kingdom",
    "China, Hong Kong Special Administrative Region": "Hong Kong"
}
energy['Country'] = energy['Country'].replace(dict_names)
```

```python
gdp = pd.read_excel('world_bank.xls', skiprows=4)
gdp.rename(columns={gdp.columns[0]: 'Country Name'}, inplace=True)

gdp_names = {
    "Korea, Rep.": "South Korea",
    "Iran, Islamic Rep.": "Iran",
    "Hong Kong SAR, China": "Hong Kong"
}
gdp['Country Name'] = gdp['Country Name'].replace(gdp_names)
gdp = pd.concat([gdp.iloc[:, [0]], gdp.iloc[:, -10:]], axis=1)

gdp.columns = ['Country Name', '2006', '2007', '2008', '2009', '2010', '2011', '2012', '2013', '2014', '2015']
gdp.head()
```

```python
ScimEn = pd.read_excel('scimagojr.xlsx')

df = pd.merge(ScimEn, energy, how='inner', on='Country')
df = pd.merge(df, gdp, how='inner', left_on='Country', right_on='Country Name')
df = df.set_index('Country')

Top15 = df.sort_values('Rank').head(15)
Top15 = Top15.drop('Country Name', axis=1)
Top15
```

```python
years = ['2006', '2007', '2008', '2009', '2010', '2011', '2012', '2013', '2014', '2015']
avgGDP = Top15[years].mean(axis=1).sort_values(ascending=False)
avgGDP
```

```python
Top15['Ratio'] = Top15['Self-citations'] / Top15['Citations']

max_country = Top15['Ratio'].idxmax()
max_value = Top15['Ratio'].max()

print(max_country, max_value)
```

```python
def answer_three():
    target = avgGDP.index[5]

    res = Top15.loc[target]['2015'] - Top15.loc[target]['2006']
    return res

answer_three()
```

```python
Top15['Population'] = Top15['Energy Supply'] / Top15['Energy Supply per Capita']

third_pop = Top15['Population'].sort_values(ascending=False).index[2]
third_pop
```

```python
Top15['Citable docs per Capita'] = Top15['Citable documents'] / Top15['Population']

correlation = Top15['Citable docs per Capita'].corr(Top15['Energy Supply per Capita'])
correlation
```

```python
ContinentDict  = {'China':'Asia', 'United States':'North America', 'Japan':'Asia',
                  'United Kingdom':'Europe', 'Russian Federation':'Europe', 'Canada':'North America',
                  'Germany':'Europe', 'India':'Asia', 'France':'Europe', 'South Korea':'Asia',
                  'Italy':'Europe', 'Spain':'Europe', 'Iran':'Asia', 'Australia':'Australia',
                  'Brazil':'South America'}

Top15['Continent'] = pd.Series(ContinentDict)

ans_seven = Top15.groupby('Continent')['Population'].agg(['size', 'sum', 'mean', 'std'])
ans_seven
```
